```python
import numpy as np
import pandas as pd
print("Numpy version:",np.__version__)
print("Pandas version:",pd.__version__)
```

```python
marks=np.array([78,85,92,67,88])
print("Marks:",marks)
print("Types:",type(marks))
```

```python
#numpy properties
print("No of Dimension=",marks.ndim)
print("Shape=",marks.shape)
print("Number of ele=",marks.size)
```

```python
#numpy operation
print("Add 5:",marks+5)
print("Multiply by 2:",marks*2)
print("Divide by 2:",marks/2)
print("Square:",marks**2)
print("Max:",marks.max())
print("Min:",marks.min())
print("Mean:",marks.mean())#sum of val / total no of val
print("Sum:",marks.sum())
```

```python
#numpy indexing
print("First marks:",marks[0])
print("Second marks:",marks[1])
print("Index 1 to 3:",marks[1:4])
print("First three:",marks[:3])
print("From index two:",marks[2:])
print("Every second element:",marks[::2])#step
```

```python
#2D Numpy Array
stu_marks=np.array([
    [78,85,92],
    [67,88,90],
    [100,99,89]
])
print(stu_marks)
print("Shape:",stu_marks.shape)
print("Mean of alll val:",stu_marks.mean())
print("Column-wise mean:",stu_marks.mean(axis=0))# operation performed downward
print("Row-wise mean:",stu_marks.mean(axis=1)) # operation row0wise
```

```python
#Pandas (Dataframe=2D tabular data structure in Pandas)
data={
    "Name":["Anu","Kiran","Rahul","Geeta","Priya"],
    "Branch":["CSE","CSE","AIML","CSE","AIML"],
    "Marks":[85,90,100,89,100]
}

df=pd.DataFrame(data)
df
```

```python
#Inscepting DataFrame
print("shape:",df.shape)
print("columns:",df.columns)
df.head()
df.tail()
df.info()
df.describe()
```

```python
#Selecting Columns
print("Marks column:",[df["Marks"]])
```

```python

#iloc= integer based location indexing
print("First Row")
print(df.iloc[0])
print("\nSecond Row")
print(df.iloc[1])
print("Fist three rows:")
print(df.iloc[0:3])
```

```python
print("first 2 rows and first 2 columns")
print(df.iloc[0:2, 0:2])
```

```python
#DataFrame silicing
print("First three rows:")
display(df.iloc[0:3])
print("Move 1 to 3")
display(df.iloc[1:4, 0:1])
```

```python
#Data Filtering
hs=df[df["Marks"]>95]
hs
print("Student From CSE:")
display(df[df["Branch"]=="AIML"])
```
