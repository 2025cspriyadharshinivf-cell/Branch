```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J","k","L","M",
                "N","O"],
    "Study_Hours":[2,3,4,5,4,6,5,3,4,5,6,4,5,30,3]
}
df=pd.DataFrame(data)
print("Original Dataset")
display(df)
```

```python
df["Study_Hours"].describe()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df["Study_Hours"])
plt.title("Boxplot of Study Hours")
plt.xlabel("Study Hours")
plt.show()
```

```python
#Detection method using IQR

#step1 find Q1&Q2

Q1 = df["Study_Hours"].quantile(0.25)
Q3 = df["Study_Hours"].quantile(0.75)

print("Q1:",Q1)
print("Q3:",Q3)
```

```python
#step2 Calculate IQR

IQR = Q3 - Q1
print("IQR:",IQR)
```

```python
#Step3 Calulate Boundaries

lb = Q1 - 1.5 * IQR
ub = Q3 + 1.5 * IQR

print("Lower Bound:",lb)
print("Upper Bound:",ub)
```

```python
#Step4 Identify IQR outlier

outlier_iqr = df[
    (df["Study_Hours"] < lb) |
    (df["Study_Hours"] > ub)
]

print("IQR Outliers:")
display(outlier_iqr)


```

```python
#Step5 Identify IQR outlier

normal_data_iqr = df[
    (df["Study_Hours"] >= lb) &
    (df["Study_Hours"] <= ub)
]

print("Normal IQR Outliers:")
display(normal_data_iqr)
```

```python
df["z_score"] = zscore(df["Study_Hours"])
print("Dataset with Z-Score:")
display(df)
```

```python
#Detection using zscore Method

z_outliers = df[
    (df["z_score"] < -3) |
    (df["z_score"] > 3)
]

print("Z-Score Outliers:")
display(z_outliers)
```

```python
print("IQR outlier")
display(outlier_iqr)

print("Z-Score outlier")
display(z_outliers)
```

```python
#Treatment1 -> Remove IQR outlier
df_without_outliers = df[
    (df["Study_Hours"] >= lb) &
    (df["Study_Hours"] <= ub)
]

print("Dataset without Outliers:")
display(df_without_outliers)
```

```python
#Treatment2 -> Capping
df_capped = df.copy()

df_capped["Study_Hours"] = df_capped["Study_Hours"].clip(
    lower=lb,
    upper=ub
)

print("Dataset with Capped Values:")
display(df_capped)
```

```python
#Final visualisation

plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
sns.boxplot(x=df["Study_Hours"])
plt.title("Before Treatment Boxplot of Study Hours")
plt.xlabel("Study Hours")

plt.subplot(1,2,2)
sns.boxplot(x=df_capped["Study_Hours"])
plt.title("After Treatment Boxplot of Study Hours")
plt.xlabel("Study Hours")

plt.tight_layout()
plt.show()
```
